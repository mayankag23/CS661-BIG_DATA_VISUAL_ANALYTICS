```python
import vtk
from vtk import *
from vtk.util import numpy_support as numpy_support


# Create a reader for .vti file 
reader = vtk.vtkXMLImageDataReader()
reader.SetFileName("mixture.vti")  # Replace "your_file.vti" with the path to your .vti file
reader.Update()


# As  plotly can not work with .vti data directlt so we need to convert it to numpy array
scalar_data = reader.GetOutput().GetPointData().GetScalars()
scalar_data_np = numpy_support.vtk_to_numpy(scalar_data)

reshaped_data=scalar_data_np.reshape(reader.GetOutput().GetDimensions()[::-1])

```

```python
# pip install plotly
import plotly.graph_objs as go
# define a function which return iso_surface plot based on iso_value
def plot_iso_surface(iso_value):
    z, y, x = np.mgrid[:reshaped_data.shape[0], :reshaped_data.shape[1], :reshaped_data.shape[2]]
    iso_surface = go.Isosurface(
        x=x.ravel(),
        y=y.ravel(),
        z=z.ravel(),
        value=reshaped_data.ravel(),
        isomin= iso_value, # setting both isomin & isomax as isovalue
        isomax= iso_value, # to make single isosurface for the given value
        surface_count=1,
        colorscale='plasma',    #colour scale is plasma
        cmin=np.min(scalar_data),
        cmax=np.max(scalar_data),
        showscale = False,
        
    )
    figure = go.Figure(data=[iso_surface])
    
    return figure

    
```

```python
# import numpy as np
import plotly.express as px

# define a function which return histogram plot based on iso_value
def plot_histogram(iso_value):
    if iso_value == 0:
#         set the entire range as given in question
        scaler_data_range = scalar_data_np 
    else:
        scaler_data_range = [x for x in scalar_data_np if (iso_value - 0.25) <= x <= (iso_value + 0.25)]
    histogram = px.histogram(x = scaler_data_range, nbins = 40)
    
    return histogram
    

    
```

#Here we are just merging two figures in a single figure and just see to check

```python
from plotly.subplots import make_subplots
import numpy as np

final_fig = go.Figure(make_subplots(rows=1, cols=2, specs=[[{'type': 'scene'}, {'type': 'histogram'}]]))
traces = []
traces.append(plot_iso_surface(0).data[0])
traces.append(plot_histogram(0).data[0])
final_fig.add_traces(traces)

# creating final layout
final_fig.update_layout(  title='Iso_Surface and Histogram Figure ',
    height=500,
    width=900,
    scene=dict(
        xaxis=dict(title='X', showticklabels=False),
        yaxis=dict(title='Y', showticklabels=False),
        zaxis=dict(title='Z', showticklabels=False)
    ),
    xaxis_title='Vertex Scalar Values',
    yaxis_title='Frequency')
```

```python
# Adding the slider
import ipywidgets as widgets

slider = widgets.FloatSlider()

slider.value = 0.00
slider.min = scalar_data.GetRange()[0]
slider.max = scalar_data.GetRange()[1]
slider.step = 0.01
slider.description='Isoval:'
slider.continuous_update=True
slider.orientation='horizontal'
slider.readout=True
slider.readout_format='.2f'

```

```python
# Adding the button
button = widgets.Button(description="Reset")
Reset_flag=0
def on_button_click(b):
    global Reset_flag
    Reset_flag=1
    slider.value = 0.00
    
# button.on_click(on_button_clicked)
```

```python
from ipywidgets import IntSlider, HBox, Button, Output
# import plotly.colors
import random

# colour_list = plotly.colors.PLOTLY_SCALES.keys()
output = Output()

flag=1
# Define the slider

def change_final_plot(change):
    value=change.new
    change.new=round(change.new,2)
    global Reset_flag
    global flag
    n = change.new

    final_fig.data[0].isomin = n
    final_fig.data[0].isomax = n

    if (n == 0.00 and flag==1) or (n == 0.0 and Reset_flag==1):
        scaler_data_range = scalar_data_np 
        Reset_flag=0
        flag=0
    else:
        scaler_data_range = [x for x in scalar_data_np if (value - 0.25) <= x <= (value + 0.25)]
    final_fig.data[1].x = scaler_data_range
    slider.continuous_update=False
    with output:
        output.clear_output(wait=True)
        final_fig.show()

# Observe changes in the slider value
slider.observe(change_final_plot, names='value')



# Display both Isoval slider and Reset button side by side
display(HBox([slider, button]))

display(output)

button.on_click(on_button_click)
```
